```python
S = input("Enter String")
count=0

for i in S:
    if i in "aeiouAEIOU":
        count +=1
print(f"vowels : {count}")
```

```python
original_list=[1,2,3,4,5]
reversed_list=[]

n=len(original_list)
while(n>0):
    reversed_list.append(original_list[n-1])
    n-=1
print("rev : ",reversed_list)
```
